**Cohort Table**


The lead source is used as the cohort dimension because the business objective is to understand which sources generate leads that are more likely to convert to FT.

The primary performance metric is FT conversion rate, supported by total FT conversions and total lead volume.

```python
cohort = (
    df_clean
    .groupby("lead_source")
    .agg(
        total_leads=("candidate_phone", "size"),
        ft_candidates=("FT_after_first_attempt", "sum")
    )
    .reset_index()
)

cohort["ft_conversion_rate"] = (
    cohort["ft_candidates"] /
    cohort["total_leads"] * 100
)

cohort = cohort.sort_values(
    "ft_conversion_rate",
    ascending=False
)

cohort
```

```python
top_3 = cohort.head(3)

top_3
```

cohort after secondary measuer

```python
upload_cohort = (
    df_clean
    .groupby("lead_source")
    .agg(
        total_leads=("lead_source", "size"),
        ft_after_upload=("FT_after_upload", "sum")
    )
    .reset_index()
)

upload_cohort["upload_ft_rate"] = (
    upload_cohort["ft_after_upload"] /
    upload_cohort["total_leads"] * 100
)

upload_cohort.sort_values(
    "upload_ft_rate",
    ascending=False
).head(3)
```

```python
top_10 = cohort.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_10["lead_source"],
    top_10["ft_conversion_rate"]
)

plt.xlabel("FT Conversion Rate (%)")
plt.ylabel("Lead Source")
plt.title("Lead Source Performance by FT Conversion Rate")

plt.gca().invert_yaxis()
plt.show()
```

Metric selection: FT conversion rate was selected as the primary metric because it measures the proportion of leads from each source that ultimately converted to FT. FT volume was used as a supporting metric to account for the scale of each lead source.

Then:

Top 3 cohorts: Based on FT conversion rate, the top-performing lead sources were X, Y and Z.

## Machine Learning — Predicting FT

The objective is to build a classification model that estimates whether a lead will result in FT.

The target variable is FT_after_first_attempt.

The model will also be used to identify the factors that are associated with the likelihood of FT.

## Feature Preparation

```python
target = "FT_after_first_attempt"

X = df_clean.drop(columns=[target]).copy()
y = df_clean[target].astype(int)
```

```python
X.info()
```

```python
id_columns = [
    "lead_id",
    "candidate_id",
    "customer_id",
    "id"
]

X = X.drop(
    columns=[c for c in id_columns if c in X.columns],
    errors="ignore"
)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numeric features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
date_features = X.select_dtypes(
    include=["datetime64[ns]"]
).columns.tolist()

date_features
```

```python
for col in date_features:
    X[col + "_year"] = X[col].dt.year
    X[col + "_month"] = X[col].dt.month
    X[col + "_day"] = X[col].dt.day

X = X.drop(columns=date_features)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "int32", "float64", "float32"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

## Preprocessing

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

##Logistic Regression with class balancing

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced"
            )
        )
    ]
)

logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)

y_prob = logistic_model.predict_proba(X_test)[:, 1]
```

## Model Evaluation

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python
model_metrics = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ],
    "Score": [
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred),
        roc_auc_score(y_test, y_prob)
    ]
})

model_metrics
```

## Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm
)

disp.plot()

plt.title("Confusion Matrix — FT Prediction")
plt.show()
```

### Confusion Matrix Interpretation

The confusion matrix evaluates how effectively the model distinguishes between leads that convert to FT and those that do not.

- True Positive: FT leads correctly predicted as FT.
- True Negative: Non-FT leads correctly predicted as non-FT.
- False Positive: Non-FT leads incorrectly predicted as FT.
- False Negative: FT leads incorrectly predicted as non-FT.

## Factor Associated with FT

```python
positive_factors = (
    feature_importance
    .sort_values("coefficient", ascending=False)
    .head(10)
)

positive_factors
```
